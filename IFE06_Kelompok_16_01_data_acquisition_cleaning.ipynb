## Transformasi Fitur

Berdasarkan hasil audit tipe data, seluruh 14 variabel prediktor pada dataset
memiliki tipe data numerik (`float64` dan `int64`). Tidak ditemukan fitur
kategorikal nominal maupun ordinal, sehingga One-Hot Encoding (OHE) dan
Ordinal Encoding tidak diterapkan. Penerapan encoding pada fitur numerik
tidak diperlukan dan dapat mengubah makna asli data.

Seluruh fitur numerik kemudian distandardisasi menggunakan `StandardScaler`
melalui `ColumnTransformer`. Standardisasi dilakukan karena setiap fitur
memiliki rentang nilai dan satuan yang berbeda, seperti temperature,
pressure, latency, throughput, dan anomaly_score. Standardisasi membuat
fitur berada pada skala yang sebanding sehingga lebih siap digunakan pada
tahap pemodelan machine learning berikutnya.

```python
numeric_features = X.columns.tolist()

print("Jumlah fitur numerik:", len(numeric_features))
print(numeric_features)
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), numeric_features)
    ]
)
```

```python
X_scaled = preprocessor.fit_transform(X)

print("Ukuran data setelah preprocessing:", X_scaled.shape)
```

```python
X_scaled[:5]
```

```python
X_scaled_df = pd.DataFrame(
    X_scaled,
    columns=numeric_features
)

X_scaled_df.head()
```

```python
dataset_clean = pd.concat(
    [X_scaled_df, y.reset_index(drop=True)],
    axis=1
)

dataset_clean.head()
```

```python
print("Jumlah baris:", dataset_clean.shape[0])
print("Jumlah kolom:", dataset_clean.shape[1])
```

```python
dataset_clean.info()
```

```python
dataset_clean.isnull().sum()
```

```python
dataset_clean["threat_level"].value_counts()
```

```python
dataset_clean.to_csv("dataset_clean.csv", index=False)

print("=== VALIDASI DATASET AKHIR ===")
print("Jumlah baris :", dataset_clean.shape[0])
print("Jumlah kolom :", dataset_clean.shape[1])
print("Missing value:", dataset_clean.isnull().sum().sum())
print("Duplikat     :", dataset_clean.duplicated().sum())

print("\nDistribusi target:")
print(dataset_clean["threat_level"].value_counts())
print("Dataset berhasil disimpan sebagai dataset_clean.csv")

files.download("dataset_clean.csv")
```

## Ringkasan Preprocessing

Dataset **Secure IoT and WSN Sensor Data** yang digunakan berasal dari Kaggle
dan terdiri dari 1.000 baris serta 15 kolom, sehingga memenuhi ketentuan
jumlah dataset yaitu 500–10.000 baris. Berdasarkan hasil audit schema,
